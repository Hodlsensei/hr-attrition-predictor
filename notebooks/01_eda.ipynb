```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_theme(style="whitegrid")

df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

print("Shape:", df.shape)
print("\nMissing Values:", df.isnull().sum().sum(), "total")
df.head()
```

```python
# How balanced is our target?
attrition_counts = df['Attrition'].value_counts()
print(attrition_counts)
print(f"\nAttrition Rate: {attrition_counts['Yes'] / len(df) * 100:.1f}%")

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Attrition', palette='Set2')
plt.title('Attrition Distribution')
plt.show()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# See all unique values in categorical columns
cat_cols = df.select_dtypes(include='object').columns.tolist()
print("Categorical columns:", cat_cols)
for col in cat_cols:
    print(f"\n{col}: {df[col].unique()}")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

cat_features = ['Department', 'JobRole', 'MaritalStatus', 
                'BusinessTravel', 'OverTime', 'Gender']

for ax, col in zip(axes.flatten(), cat_features):
    ct = df.groupby(col)['Attrition'].value_counts(normalize=True).unstack()
    ct['Yes'].sort_values().plot(kind='bar', ax=ax, color='salmon')
    ax.set_title(f'Attrition Rate by {col}')
    ax.set_ylabel('Attrition Rate')
    ax.set_xlabel('')
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

num_features = ['Age', 'MonthlyIncome', 'DistanceFromHome',
                'YearsAtCompany', 'TotalWorkingYears', 'YearsWithCurrManager']

for ax, col in zip(axes.flatten(), num_features):
    df.boxplot(column=col, by='Attrition', ax=ax)
    ax.set_title(f'{col} vs Attrition')
    ax.set_xlabel('Attrition')

plt.suptitle('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))
num_df = df.select_dtypes(include='number')
corr = num_df.corr()
sns.heatmap(corr, annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```
